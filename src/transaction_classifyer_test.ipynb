```python
df = pd.read_csv("../data/processed/period-202410.csv")
df.head()
```

```python
def classify_text(text):
    try:
        completion = client.chat.completions.create(
            model="gpt-4o-mini",  # Replace with the desired model
            messages=[
                {"role": "system", "content": "You are an expert in categorizing financial transactions into predefined categories such as Food, Shopping, Groceries, Entertainment, Utilities, Travel, and Others."},
                {
                    "role": "user", 
                    "content": f"Categorize the following transaction description into one of these categories: Food, Shopping, Groceries, Entertainment, Utilities, Travel, Others. Answer with only the Category and nothing else Description: {text}"
                }
            ]
        )
        # Extract and return the classification from the completion
        return completion.choices[0].message.content
    except Exception as e:
        print(f"Error classifying text: {e}")
        return "Error"
```

```python
# Apply the function to the 'beskrivning' column and create a new column
df.head()
```

```python
df['classification'] = df['Beskrivning'].apply(classify_text)
```
